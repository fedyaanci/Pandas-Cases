```python
import pandas as pd

def getDataframeSize(players: pd.DataFrame) -> List[int]:
    return [players.shape[0], players.shape[1]]
```

```python
def createDataframe(student_data: List[List[int]]) -> pd.DataFrame:
    return pd.DataFrame(data=student_data, columns=['student_id', 'age'])
```

```python
def selectFirstRows(employees: pd.DataFrame) -> pd.DataFrame:
    return employees.head(3)
    
```

```python
def selectData(students: pd.DataFrame) -> pd.DataFrame:
    return students.loc[students['student_id'] == 101, ['name', 'age']]
```

```python
def createBonusColumn(employees: pd.DataFrame) -> pd.DataFrame:
    return employees.assign(bonus=lambda x: x['salary']*2)
```

```python
def dropDuplicateEmails(customers: pd.DataFrame) -> pd.DataFrame:
    return customers.drop_duplicates(subset='email', keep='first')
```

```python
def dropMissingData(students: pd.DataFrame) -> pd.DataFrame:
    return students.dropna(subset='name')
```

```python
def modifySalaryColumn(employees: pd.DataFrame) -> pd.DataFrame:
    return employees.assign(salary= lambda x: x['salary']*2)
    
```

```python
def renameColumns(students: pd.DataFrame) -> pd.DataFrame:
    return students.rename(columns={'id': 'student_id', 'first': 'first_name', 'last':'last_name','age':'age_in_years'})
```

```python
def changeDatatype(students: pd.DataFrame) -> pd.DataFrame:
    return students.astype({'grade': 'Int64'})
```

```python
def fillMissingValues(products: pd.DataFrame) -> pd.DataFrame:
    return products.fillna({'quantity': 0})
```

```python

def concatenateTables(df1: pd.DataFrame, df2: pd.DataFrame) -> pd.DataFrame:
    return pd.concat([df1, df2], axis=0)
```

```python
def pivotTable(weather: pd.DataFrame) -> pd.DataFrame:
    return pd.pivot_table(data=weather,index='month', columns='city', values='temperature')
```

```python
def meltTable(report: pd.DataFrame) -> pd.DataFrame:
    return (report
            .melt(id_vars='product', var_name='quarter', value_name='sales')
            .sort_values('quarter', kind='stable')
            .reset_index(drop=True))
```

```python
def findHeavyAnimals(animals: pd.DataFrame) -> pd.DataFrame:
    return animals.loc[animals['weight'] > 100].sort_values(by =['weight'], ascending=False)[['name']]
```
