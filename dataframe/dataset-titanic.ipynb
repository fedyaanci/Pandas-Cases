```python
round(list_ages.std(), 2)
#std()- Стандартное отклонение (разброс)  это число, которое показывает,
#  насколько сильно данные «разбросаны» вокруг среднего значения
#  Математически:
# 1) Находим среднее: среднее = сумма / количество
# 2) Считаем, насколько каждое значение отклоняется от среднего: x - среднее
# 3) Возводим в квадрат (чтобы минусы не мешали): (x - среднее)^2
# 4) Находим среднее этих квадратов - это дисперсия
# 5) Извлекаем квадратный корень - получаем стандартное отклонение (чтобы вернуться к исходным единицам)
```

```python
round(list_ages.var(), 2) # дисперсия
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Считаем матрицу
corr = data.corr(numeric_only=True)

# Рисуем тепловую карту
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0)
plt.show()
```

Красный = положительная корреляция (растут вместе)
Синий = отрицательная корреляция (одно растёт, другое падает)
Белый/светлый = слабая или нулевая связь

Корреляция ≠ причинно-следственная связь
Выбросы сильно искажают результат
Работает только с числами
